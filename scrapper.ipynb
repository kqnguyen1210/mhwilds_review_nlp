```python
# Libraries
import requests
import pandas as pd
import time
```

```python
# Steam App ID for MHWilds
app_id = 2246340

# Number of reviews to pull
target_reviews = 1000
all_reviews = []
```

```python
cursor = '*' 

print(f"Initiating mass download for App ID: {app_id}...")

while len(all_reviews) < target_reviews:
    params = {
        'json': 1,
        'filter': 'recent',      # Get the newest reviews
        'language': 'english',   # Keep it to one language for the NLP model
        'num_per_page': 100,     # Max allowed by Steam
        'cursor': cursor
    }
    
    url = f"https://store.steampowered.com/appreviews/{app_id}"
    response = requests.get(url, params=params)
    
    if response.status_code != 200:
        print("Server blocked us! Stopping early.")
        break
        
    data = response.json()
    batch = data.get('reviews', [])
    
    # If the batch is empty, we ran out of reviews
    if len(batch) == 0:
        print("No more reviews found.")
        break
        
    all_reviews.extend(batch)
    
    cursor = data.get('cursor')
    
    print(f"Secured {len(all_reviews)} reviews so far...")
    
    time.sleep(2) 

# Convert the raw JSON data into a clean Pandas DataFrame
df = pd.DataFrame(all_reviews)

# Save it to a CSV file (Don't forget to add this to your .gitignore!)
df.to_csv("mh_wilds_reviews.csv", index=False)

print("Data secured in mh_wilds_reviews.csv")
```
